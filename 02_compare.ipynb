```python
import pandas as pd, glob, os

pd.set_option("display.max_colwidth", None)
frames = []
for f in sorted(glob.glob("results/*.csv")):
    df = pd.read_csv(f)
    df.insert(0, "approach", os.path.basename(f).replace(".csv", ""))
    frames.append(df)

all_res = pd.concat(frames, ignore_index=True)
all_res.sort_values("ndcg@k", ascending=False).round(4)
```

```python
best = all_res.loc[all_res.groupby("approach")["ndcg@k"].idxmax()]
best.sort_values("ndcg@k", ascending=False).round(4)
```
