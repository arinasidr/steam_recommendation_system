```python
import sys; sys.path.append("src")
import numpy as np, matplotlib.pyplot as plt
from common import load_data

X, test_sets, maps = load_data()
```

```python
per_game = np.asarray((X > 0).sum(axis=0)).ravel()
plt.hist(per_game, bins=50, log=True)
plt.xlabel("число пользователей у игры"); plt.ylabel("число игр (log)")
plt.title("Распределение популярности игр"); plt.show()
print("топ-10 игр дают %.1f%% всех взаимодействий" %
      (100 * np.sort(per_game)[-10:].sum() / per_game.sum()))
```

```python
per_user = np.asarray((X > 0).sum(axis=1)).ravel()
plt.hist(per_user, bins=50, log=True)
plt.xlabel("игр на пользователя"); plt.title("Активность пользователей"); plt.show()
```

```python
hours = np.expm1(X.data - 1)
plt.hist(hours, bins=60, log=True)
plt.xlabel("часы в игре"); plt.title("Распределение часов"); plt.show()
print("медиана часов:", np.median(hours), " максимум:", hours.max())
```
